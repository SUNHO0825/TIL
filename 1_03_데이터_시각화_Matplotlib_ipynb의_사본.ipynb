**계열별 선형 차트**

```python
dataset=[[25, 100], [52, 256], [38, 152], [32, 140], [25, 150], [45, 183], [40, 175], [55, 203],[28, 152], [42, 198]]
actual=[j[1] for j in dataset]
 # 리스트 내포(list comprehension)를 사용, dataset에서 두 번째 값(실제 값)만을 추출하여 actual 리스트에 저장
 # 두번째 값 j[1]
 # [100, 256, 152, 140, 150, 183, 175, 203, 152, 198]

predicted=[115, 242, 176, 148, 115, 209, 185, 256, 129, 195]
error=[(x - y) for x, y in zip(actual, predicted)]

xs=[i for i, _ in enumerate(error)]
plt.plot(xs, actual, 'g-', label='x') # 녹색 실선
plt.plot(xs, predicted, 'r-.', label='y') # 빨간색 점선
plt.plot(xs, error, 'b', label='error') # 파란색 점선
plt.legend(loc=9) # 범례 표시: loc 숫자로 위치 설정
plt.xlabel('mode')
plt.title('actual vs predicted')
plt.show()
```

```python
import cv2

img_gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# 흑백 이미지 출력
plt.imshow(img_gray, cmap="gray")
plt.axis("off")
plt.title("gray image")
plt.show()
```

```python
def rotate_image(image, angle):
    """이미지를 특정 각도로 회전"""
    h, w = image.shape[:2]
    center = (w // 2, h // 2)
    matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
    rotated = cv2.warpAffine(image, matrix, (w, h))
    return rotated

# 45도 회전
img_rotated = rotate_image(img, 45)

# 출력
plt.imshow(img_rotated)
plt.axis("off")
plt.title("45...")
plt.show()
```

```python
# 가우시안 블러 적용
img_blur = cv2.GaussianBlur(img, (15, 15), 0)

# 출력
plt.imshow(img_blur)
plt.axis("off")
plt.title("blur")
plt.show()
```

```python
def adjust_brightness(image, factor):
    """이미지 밝기 조절 (factor > 1: 밝게, factor < 1: 어둡게)"""
    img_float = image.astype(np.float32) * factor
    img_float = np.clip(img_float, 0, 255)  # 값 범위를 0~255로 제한
    return img_float.astype(np.uint8)

# 밝기 증가
img_bright = adjust_brightness(img, 1.5)

# 출력
plt.imshow(img_bright)
plt.axis("off")
plt.title("brightness")
plt.show()
```
