1-2. 문자열을 숫자로 변환하기

```python
print(df_house.columns)
```

```python
import pandas as pd
houses = []

for index, row in df_house.iterrows():
    if (
        pd.isna(row["ID"])
        or pd.isna(row["size"])
        or pd.isna(row["year"])
        or pd.isna(row["price"])
    ):
        print(index, "필수 값이 비어 있어 제외")
        continue

    try:
        house_id = int(row["ID"])
        size = int(row["size"])
        year = int(row["year"])
        price = float(row["price"])
    except (ValueError, OverflowError):
        print(index, "숫자로 변환할 수 없어 제외")
        continue

    house = {"ID": house_id, "size": size, "year": year, "price": price}
    houses.append(house)

print("정상 저장 행 수:", len(houses))
print("제외 행 수:", len(df_house) - len(houses))
```

1-3. 조건에 맞는 주택 찾기

```python
house_test=[
    {"ID": 101, "year": 5, "price": 3},   # size 키 없음
    {"ID": 102, "size": 35, "year": 5, "price": None},   # 값 없음
    {"ID": 103, "size": "오류", "year": 5, "price": 3},   # 숫자로 변환할 수 없음
    {"ID": 104, "size": 35, "year": 5, "price": 3}  # 정상 데이터
]

count=0
count_and=0

for house in houses:
    if house['size']<=0 or house['price']<=0:
        continue

    if house['size']>=30 or house['price']<=4:
        print('or 조건에 만족하는 수:', house['ID'])
        count+=1

    if house['size']>=30 and house['price']<=4:
        print('and 조건에 만족하는 수:', house['ID'])
        count_and+=1

print('한 조건 이상을 만족하는 집 수:', count)
print('두 조건을 만족하는 맞는 짐 수:', count_and)
```

```python
def unit_price(price, size):
    try:
        price=float(price)
        size=float(size)
    except ValueError:
        print('값 오류: 숫자를 변환할 수 없는 문자입니다.')
        return None
    except TypeError:
        print('자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.')
        return None

    if size <= 0:
        print('크기 오류: 크기는 0보다 커야 합니다.')
        return None

    if price < 0:
        print('가격 오류: 가격은 0 이상이어야 합니다.')
        return None

    return price/size
```

```python
print(unit_price(4, 40))
```

```python
print(unit_price('4', '40'))
```

```python
print(unit_price(4, 0))
```

```python
print(unit_price(-4, 40))
```

```python
print(unit_price("오류", 40))
```

```python
print(unit_price(None, 40))
```

```python
print(unit_price(6, 30))
```

1-5. 목적에 맞는 자료구조 선택하기

```python
# 리스트는 ID를 추가한 순서대로 저장하고, 중복 ID도 그대로 보관할 수 있다.
# 집합은 중복을 제거하고 원래 입력 순서를 보장하지 않는다.
house_ids = []

for house in houses:
    house_ids.append(house["ID"])

print(house_ids)
```

```python
# 집합은 같은 year 값이 여러 번 나와도 한 번만 저장한다.
# 리스트는 같은 값을 추가하면 중복을 그대로 저장한다.
house_years = set()

for house in houses:
    house_years.add(house["year"])

print(house_years)
```

```python
# 딕셔너리는 주택 ID를 키로 사용해 해당 가격을 바로 찾을 수 있다.
# 리스트의 인덱스는 주택 ID가 아니라 저장된 위치를 나타낸다.
price_by_id = {}

for house in houses:
    price_by_id[house["ID"]] = house["price"]

print(price_by_id[3])
```
